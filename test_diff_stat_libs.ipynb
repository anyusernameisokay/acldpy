# Example 1: Penguins

```python
# 1) Usage with pingouin
penguins = pg.read_dataset("penguins")
penguins_tk_results = penguins.pairwise_tukey(dv='body_mass_g', between='species')
group_1_col, group_2_col, pvals = list(penguins_tk_results['A']), list(penguins_tk_results['B']), list(penguins_tk_results['p-tukey'])
final_letters = calc_cld.run_clc(group_1_col, group_2_col, pvals)
print(final_letters)
```

```python
# 2) Usage with with scipy tukey-kramer
# Skipt this for know, it is a little weird.
from scipy.stats import tukey_hsd
xx = [penguins.loc[penguins['species'] == sp, 'body_mass_g'].dropna().astype(int).tolist()
    for sp in penguins['species'].unique()]
scipy_tk_results = tukey_hsd(*xx)
print(scipy_tk_results)
```

```python
# Using stats models
from statsmodels.stats.multicomp import pairwise_tukeyhsd
data = penguins[['body_mass_g', 'species']].dropna()
statsmodels_tk_results = pairwise_tukeyhsd(data['body_mass_g'], data['species'])
group_1_col, group_2_col, pvals = list(statsmodels_tk_results['group1']), list(statsmodels_tk_results['group2']), list(statsmodels_tk_results['p-adj'])
final_letters = calc_cld.run_clc(group_1_col, group_2_col, pvals)
```
